```python
# import libraries
import pandas as pd
import numpy as np
```

```python
#load dataset
df = pd.read_csv(r"C:\Users\Chaithanya\Desktop\wrangling\student_data.csv")
print("\nfirst 5 rows of dataset\n")
print(df.head())
```

```python
#data immersion
print("\ndataset information\n")
print(df.info())

print("\nstatistical summary\n")
print(df.describe())

print("\ncolumn names\n")
print(df.columns)
```

```python
#identify missing values
print("\nmissing values in each column\n")
print(df.isnull().sum())

print("\ntotal missing values\n")
print(df.isnull().sum().sum())
```

```python
#identify duplicate values
duplicates = df.duplicated().sum()

print("\nnumber of duplicate rows\n")
print(duplicates)

print("\nduplicate rows\n")
print(df[df.duplicated()])
```

```python
# handle missing values
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nmissing values after handling\n")
print(df.isnull().sum())
```

```python
# remove duplicates
df = df.drop_duplicates()
print("\nduplicate rows after removal\n")
print(df.duplicated().sum())
```

```python
#data transformation
df.columns = df.columns.str.lower()
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    df[col] = df[col].str.strip()

for col in categorical_cols:
    df[col] = df[col].str.upper()

if 'marks' in df.columns:
    df['percentage'] = (df['marks'] / 500) * 100

print("\ntransformed dataset\n")
print(df.head())
```

```python
# outlier detection

Q1 = df['marks'].quantile(0.25)
Q3 = df['marks'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

outliers = df[(df['marks'] < lower_limit) | (df['marks'] > upper_limit)]

print("\noutliers in marks column\n")
print(outliers)
```

```python
##save cleaned dataset
output_path = r"C:\Users\Chaithanya\OneDrive\New Folder\student_data.csv"
df.to_csv(output_path, index=False)
print("\nDATA CLEANING COMPLETED\n")
print("Cleaned Dataset Saved Successfully!")
print("Saved Location:", output_path)
```
